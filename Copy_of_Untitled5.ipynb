```python
# Exp 1: Highest Marks for each Subject in Bar Graph
import matplotlib.pyplot as plt
subjects=["Maths","Physics","Chemestry","DataScience"]
marks=[85,95,75,95]
plt.bar(subjects,marks,color="steelblue")
plt.title("Marks in each Subject")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
print("Highest Marks in Subject:",subjects[marks.index(max(marks))])
```

```python
# Exp 3: Favpurite Programming Language using Pie Chart in Python
import matplotlib.pyplot as plt
langs=["Python","Java","C","JavaScript"]
students=[40,25,15,20]
plt.pie(students,labels=langs,autopct="%1.1f%%")
plt.title("Favourite Programming Language")
plt.show()
```

```python
# Exp 5: Employee Salary Distribution using Histogram
import matplotlib.pyplot as plt
salaries=[15,18,20,22,25,25,28,30,32,35,38,40,45,48,50]
plt.hist(salaries,bins=5,edgecolor="black")
plt.title("Employee Salary Distribution")
plt.xlabel("Salary in (Rs.thousands)")
plt.ylabel("Frequency")
plt.show()
```

```python
# Exp 7: Daily Sales using Line chart
import matplotlib.pyplot as plt
days=["Mon","Tue","Wed","Thur","Fri","Sat","Sun"]
sales=[500,250,360,841,462,355,795]
plt.plot(days,sales,marker="o")
plt.title("Daily Shop Sales")
plt.xlabel("Day")
plt.ylabel("Sales (Rs.)")
plt.show()
print("Highest:",days[sales.index(max(sales))])
```

```python
# Exp 15 Sales by category, Quantity using bar ,pie and scatter plot
import matplotlib.pyplot as plt
category={
    "Electronics":5451522,
    "Furniture":845122,
    "Clothing":266661,
    "Books":26615,
    "Electronics":48166,
    "Furniture":7966
}
region={
    "South":8455,
    "North":4515,
    "West":1516,
    "East":6963,
    "North":3664,
    "South":6261
}
sales=[5500,35000,25000,12000,40000,48121]
quantity = [20, 10, 35, 30, 15, 12]

fig,ax=plt.subplots(1, 3, figsize=(15,4))
ax[0].bar(category.keys(), category.values())
ax[0].set_title("Sales by Category")
ax[1].pie(region.values(), labels=region.keys(), autopct="%1.1f%%")
ax[1].set_title("Sales by Region")
ax[2].scatter(quantity, sales)
ax[2].set_title("Sales by Quantity")
ax[2].set_xlabel("Quantity")
ax[2].set_ylabel("Sales")
plt.tight_layout()
plt.show()
print("Best category: ",max(category,key=category.get))
```

```python
import matplotlib.pyplot as plt
products=["Laptop","Mobile","HeadPhones","Monitor","keyboard","Printer","Tablet"]
sales=[120000,95000,55000,30000,18000,45000,70000]
quantity=[12,25,40,15,30,10,20]

fig,ax=plt.subplots(1, 3, figsize=(15,4))
ax[0].bar(products,sales)
ax[0].set_title("Product Sales")
ax[0].tick_params(axis="x",rotation=45)

ax[1].pie(sales,labels=products,autopct="%1.1f%%")
ax[1].set_title("Percentage Contibution")

ax[2].scatter(quantity,sales)
ax[2].set_title("Quantity vs Sales")
ax[2].set_xlabel("Quantity")
ax[2].set_ylabel("Sales")
plt.tight_layout()
plt.show()

print("Most Sold :",products[quantity.index(max(quantity))])
print("Highest Revenue :",products[sales.index(max(sales))])
```
